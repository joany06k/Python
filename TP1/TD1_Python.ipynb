````python
def retirer_negatifs(nombres):
    """Renvoie une nouvelle liste sans les nombres négatifs.
    La liste reçue ne doit pas être modifiée."""
    for n in nombres:
        if n < 0:
            nombres.remove(n)
    return nombres


def inscrire(nom, groupe=[]):
    """Ajoute nom au groupe et renvoie le groupe.
    Si aucun groupe n'est donné, on part d'un groupe vide."""
    groupe.append(nom)
    return groupe


def grille_vide(lignes, colonnes):
    """Renvoie une grille lignes x colonnes remplie de 0,
    dont chaque case peut être modifiée indépendamment."""
    return [[0] * colonnes] * lignes
````

````python
l = [-1, -2, 3, -4, 5]
print(retirer_negatifs(l))  # attendu : [3, 5]
print(l)                    # attendu : [-1, -2, 3, -4, 5] (inchangée)

print(inscrire("Ada"))      # attendu : ['Ada']
print(inscrire("Alan"))     # attendu : ['Alan']

g = grille_vide(3, 2)
g[0][0] = 1
print(g)                    # attendu : [[1, 0], [0, 0], [0, 0]]
````

Toutes les opérations n'ont pas le même coût. Tester `x in liste` oblige Python à parcourir la liste élément par élément : le temps est proportionnel à sa longueur. Tester `x in ensemble` (ou `x in dictionnaire`) est quasi immédiat, quelle que soit la taille.

Conséquence : deux boucles imbriquées sur une liste de `n` éléments font environ `n × n` opérations, ce qui devient vite très lent. Un seul parcours, en mémorisant ce qu'on a déjà vu dans un ensemble, en fait environ `n`.

On peut mesurer un temps d'exécution avec le module `time` :

```python
import time
debut = time.perf_counter()
# ... code à mesurer ...
print(time.perf_counter() - debut, "secondes")
```

`sorted` accepte un paramètre `key` : une fonction qui indique, pour chaque élément, la valeur selon laquelle trier. Une `lambda` permet d'écrire cette fonction en une ligne. Si la clé est un tuple, Python trie selon le premier élément, puis selon le second en cas d'égalité.

```python

sorted(eleves, key=lambda e: e[1])             # tri par note croissante
sorted(eleves, key=lambda e: e[1], reverse=True)  # tri par note décroissante
```

Astuce : pour trier par ordre décroissant sur un critère numérique seulement, on peut utiliser son opposé, `-e[1]`.
